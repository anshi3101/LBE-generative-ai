Custom Tools

```python
from langchain_core.tools import tool
```

```python
# Step 1 - Create a function

def multiply(a,b):
    """Multiply two numbers""" #add docstring here
    return a*b
```

```python
# Step 2 - add type hints

def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a*b
```

```python
#Step 3 - Add tool decorator
@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a*b
```

```python
result = multiply.invoke({"a": 3, "b":5})
```

```python
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
print(multiply.args_schema.model_json_schema())
```

Method 2 - Using Structured Tool

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel,Field
```

```python
class MultiplyInput(BaseModel): #pydantic class
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
```

```python
def multiply_func(a: int, b: int) -> int:
    return a * b
```

```python
multiply_tool = StructuredTool.from_function(
    func = multiply_func,
    name = "multiply",
    description="Multiply two numbers",
    args_schema=MultiplyInput
)
```

```python
result= multiply_tool.invoke({'a':3,'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)
```

Method 3 - Using BaseTool Class

```python
from langchain_core.tools import BaseTool
from typing import Type
```

```python
# arg schema using pydantic

class MultiplyInput(BaseModel):
    a: int = Field(required=True,description="the first numberr to add")
    b: int = Field(required=True,description="the second numberr to add")
```

```python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"
    
    args_schema: Type[BaseModel] = MultiplyInput
    
    def _run(self, a: int, b: int) -> int: #can create async version of tool
        return a * b
```

```python
multiply_tool = MultiplyTool()
```

```python
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)
```
